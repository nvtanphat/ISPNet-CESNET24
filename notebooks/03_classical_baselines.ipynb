# Classical Baselines — CESNET-TimeSeries24

Notebook này implement và evaluate 5 classical baselines trên preprocessed panel data.
Kết quả dùng làm **lower bound** khi so sánh với DL models.

## Baselines
| Model | Mô tả | Lý do chọn |
|---|---|---|
| **Mean** | Per-entity train mean | Absolute lower bound |
| **SN-24** | Seasonal Naive, lag-24h | Daily seasonality (ACF lag-24=0.68) |
| **SN-168** | Seasonal Naive, lag-168h | Weekly seasonality (ACF lag-168=0.58) |
| **MA-24** | Moving Average 24h | Smoothed daily baseline |
| **MA-168** | Moving Average 168h | Smoothed weekly baseline |

## Key findings từ preprocessed-statistics-report.md
- **Weekday p50**: 143M bytes vs **Weekend p50**: 32M bytes → 4.4× drop → SN-168 cần bắt được pattern này
- **Holiday p50**: 33M bytes → tương đương weekend
- **Outage obs_rate**: 0.983 vs normal 0.989 → cần flag riêng
- **n_bytes_scaled std train**: 0.738 vs test: 1.027 → distribution shift nhẹ
- **Gap >168h**: 0.51% rows → những entity này SN-168 bị NaN, phải fallback

## Evaluation Protocol
- Test set: `id_time >= 2688` (~24 weeks)
- Chỉ tính trên `observed_mask == 1`
- Inverse-transform về raw bytes trước khi tính metric
- Metrics: RMSE (raw), SMAPE (%), R² (raw)

```python
# ── Shared evaluation function (dùng cho mọi model trong dự án) ───
def evaluate(y_true: np.ndarray, y_pred: np.ndarray,
             observed_mask: np.ndarray = None) -> dict:
    """Compute RMSE / SMAPE / R² on raw bytes scale.

    Both y_true and y_pred must be inverse-transformed to raw bytes.
    observed_mask: if provided, evaluate only on mask==1 slots.
    """
    if observed_mask is not None:
        m = observed_mask.astype(bool)
        y_true = y_true[m]
        y_pred = y_pred[m]

    if len(y_true) == 0:
        return {"RMSE": np.nan, "SMAPE": np.nan, "R2": np.nan, "n_obs": 0}

    residuals = y_true - y_pred
    rmse  = float(np.sqrt(np.mean(residuals ** 2)))
    denom = np.abs(y_true) + np.abs(y_pred) + 1e-8
    smape = float(100 * np.mean(2 * np.abs(residuals) / denom))
    ss_res = np.sum(residuals ** 2)
    ss_tot = np.sum((y_true - y_true.mean()) ** 2)
    r2    = float(1 - ss_res / ss_tot) if ss_tot > 0 else np.nan

    return {"RMSE": rmse, "SMAPE": smape, "R2": r2, "n_obs": int(len(y_true))}


def inverse_transform_raw(scaled_vals: np.ndarray, scaler) -> np.ndarray:
    """RobustScaler → log1p space → expm1 → raw bytes."""
    log1p_vals = scaler.inverse_transform(scaled_vals.reshape(-1, 1)).ravel()
    return np.maximum(np.expm1(log1p_vals), 0.0)


print("evaluate() và inverse_transform_raw() defined.")
```

## 2. Baseline Prediction Functions

Tất cả baselines hoạt động trên **log1p space** (giống model DL sau này).
Sau đó inverse-transform về raw bytes để evaluate.
NaN trong log1p (unobserved) được xử lý bằng forward-fill rồi fallback về train mean.

```python
def make_predictions_entity(series_log1p: pd.Series,
                             observed_mask: pd.Series,
                             train_end: int = TRAIN_END) -> dict:
    """Generate all baseline predictions for one entity.

    Input: log1p series (NaN where unobserved), indexed by id_time.
    Returns: dict of model_name → np.ndarray (same index as series_log1p).

    Prediction is in log1p space; caller inverse-transforms to bytes.
    """
    n = len(series_log1p)
    s = series_log1p.values.copy().astype(np.float32)   # NaN where unobserved

    # Train mean (only observed train slots)
    train_mask = (series_log1p.index < train_end) & (observed_mask.values == 1)
    train_mean = float(s[train_mask].mean()) if train_mask.any() else 0.0

    # Filled series for lag/rolling: forward-fill then fill remaining with train mean
    s_filled = pd.Series(s).ffill().fillna(train_mean).values

    preds = {}

    # ── Mean ─────────────────────────────────────────────────────────
    preds["Mean"] = np.full(n, train_mean, dtype=np.float32)

    # ── Seasonal Naive 24h (lag-24) ───────────────────────────────────
    sn24 = np.full(n, np.nan, dtype=np.float32)
    sn24[24:] = s_filled[:-24]
    sn24[:24] = train_mean
    preds["SN-24"] = sn24

    # ── Seasonal Naive 168h (lag-168) ────────────────────────────────
    sn168 = np.full(n, np.nan, dtype=np.float32)
    sn168[168:] = s_filled[:-168]
    sn168[:168] = train_mean
    preds["SN-168"] = sn168

    # ── Moving Average 24h ────────────────────────────────────────────
    s_filled_series = pd.Series(s_filled)
    ma24 = s_filled_series.shift(1).rolling(24, min_periods=1).mean().values.astype(np.float32)
    preds["MA-24"] = ma24

    # ── Moving Average 168h ───────────────────────────────────────────
    ma168 = s_filled_series.shift(1).rolling(168, min_periods=1).mean().values.astype(np.float32)
    preds["MA-168"] = ma168

    return preds


BASELINE_NAMES = ["Mean", "SN-24", "SN-168", "MA-24", "MA-168"]
print(f"Baselines defined: {BASELINE_NAMES}")
```

## 3. Run All Baselines on Institutions

```python
all_results = []   # list of dicts: entity × model × split metrics

for eid, grp in tqdm(panel.groupby("id_institution"), desc="Entities"):
    grp = grp.set_index("id_time").sort_index()

    series_log1p  = grp[f"{TARGET}_log1p"]
    obs_mask_full = grp["observed_mask"]

    preds_log1p = make_predictions_entity(series_log1p, obs_mask_full)

    test_mask   = (grp.index >= VAL_END)
    outage_mask = ((grp.index >= OUTAGE_START_ID) & (grp.index <= OUTAGE_END_ID))
    full_obs    = obs_mask_full.values

    # Ground truth: use raw n_bytes column (already float64, NaN where unobserved)
    true_raw = np.where(full_obs == 1, grp[TARGET].values.astype(np.float64), np.nan)

    for model_name, pred_log1p in preds_log1p.items():
        # Baselines predict in log1p space → expm1 directly (no scaler needed)
        pred_raw = np.maximum(np.expm1(pred_log1p.astype(np.float64)), 0.0)

        test_obs   = test_mask & (full_obs == 1)
        out_obs    = outage_mask & (full_obs == 1)
        normal_obs = test_mask & ~outage_mask & (full_obs == 1)

        m_test   = evaluate(true_raw[test_obs],   pred_raw[test_obs])
        m_outage = evaluate(true_raw[out_obs],    pred_raw[out_obs])
        m_normal = evaluate(true_raw[normal_obs], pred_raw[normal_obs])

        all_results.append({
            "id_institution": eid,
            "model"         : model_name,
            "test_RMSE"     : m_test["RMSE"],
            "test_SMAPE"    : m_test["SMAPE"],
            "test_R2"       : m_test["R2"],
            "test_n_obs"    : m_test["n_obs"],
            "normal_RMSE"   : m_normal["RMSE"],
            "normal_SMAPE"  : m_normal["SMAPE"],
            "normal_R2"     : m_normal["R2"],
            "outage_RMSE"   : m_outage["RMSE"],
            "outage_SMAPE"  : m_outage["SMAPE"],
            "outage_R2"     : m_outage["R2"],
            "outage_n_obs"  : m_outage["n_obs"],
        })

df_results = pd.DataFrame(all_results)
print(f"Results shape: {df_results.shape}")
df_results.head(5)
```

## 4. Aggregate Results Table

Tổng hợp theo **median across entities** (robust hơn mean với outlier entities).

```python
def aggregate_results(df: pd.DataFrame, prefix: str = "test") -> pd.DataFrame:
    """Aggregate per-entity results: median, mean, std across entities.
    Entities with n_obs==0 are excluded (no test observations).
    """
    n_obs_col = f"{prefix}_n_obs" if f"{prefix}_n_obs" in df.columns else "test_n_obs"
    df = df[df[n_obs_col] > 0] if n_obs_col in df.columns else df
    cols = [f"{prefix}_RMSE", f"{prefix}_SMAPE", f"{prefix}_R2"]
    agg = (
        df.groupby("model")[cols]
          .agg(["median", "mean", "std"])
    )
    agg.columns = ["_".join(c) for c in agg.columns]
    return agg.loc[BASELINE_NAMES]


print("=" * 65)
print("TEST SET — Aggregate across 283 institutions (raw bytes scale)")
print(f"  (entities with 0 test obs excluded)")
print("=" * 65)
agg_test = aggregate_results(df_results, "test")
display_cols_test = [c for c in agg_test.columns if "median" in c or "mean" in c]
print(agg_test[display_cols_test].to_string())

print("\n" + "=" * 65)
print("OUTAGE SLICE — Sensor probe failure (May 20 – Jun 10, 2024)")
print("=" * 65)
agg_outage = aggregate_results(df_results, "outage")
display_cols_outage = [c for c in agg_outage.columns if "median" in c or "mean" in c]
print(agg_outage[display_cols_outage].to_string())
```

## 6. Weekend / Holiday / Outage Breakdown

Từ report: `n_bytes_p50` drop 4.4× weekday→weekend.
SN-168 bắt được pattern này; Mean/MA-24 thì không → kỳ vọng SMAPE của Mean và MA-24 cao hơn trên weekend.

```python
def run_segment_evaluation(panel: pd.DataFrame, scalers: dict,
                           segment_col: str, segment_val: int) -> pd.DataFrame:
    """Evaluate baselines on a specific segment (e.g. weekends only)."""
    rows = []
    for eid, grp in panel.groupby("id_institution"):
        grp   = grp.set_index("id_time").sort_index()
        preds = make_predictions_entity(grp[f"{TARGET}_log1p"], grp["observed_mask"])

        seg_test = (
            (grp.index >= VAL_END) &
            (grp["observed_mask"] == 1) &
            (grp[segment_col] == segment_val)
        ).values

        true_raw = np.where(grp["observed_mask"].values == 1,
                            grp[TARGET].values.astype(np.float64), np.nan)

        for model_name, pred_log1p in preds.items():
            pred_raw = np.maximum(np.expm1(pred_log1p.astype(np.float64)), 0.0)
            m = evaluate(true_raw[seg_test], pred_raw[seg_test])
            rows.append({"model": model_name,
                         "SMAPE": m["SMAPE"], "R2": m["R2"],
                         "RMSE_GB": m["RMSE"] / 1e9})

    return pd.DataFrame(rows).groupby("model").median().loc[BASELINE_NAMES]


print("Weekend segment:")
seg_weekend = run_segment_evaluation(panel, scalers, "is_weekend", 1)
print(seg_weekend[["SMAPE", "R2"]].round(4).to_string())

print("\nWeekday segment:")
seg_weekday = run_segment_evaluation(panel, scalers, "is_weekend", 0)
print(seg_weekday[["SMAPE", "R2"]].round(4).to_string())
```

```python
# ── Summary: Weekday vs Weekend vs Outage ────────────────────────────
seg_outage_rows = []
for eid, grp in panel.groupby("id_institution"):
    grp   = grp.set_index("id_time").sort_index()
    preds = make_predictions_entity(grp[f"{TARGET}_log1p"], grp["observed_mask"])

    outage_obs = (
        (grp.index >= OUTAGE_START_ID) & (grp.index <= OUTAGE_END_ID) &
        (grp["observed_mask"] == 1)
    ).values

    true_raw = np.where(grp["observed_mask"].values == 1,
                        grp[TARGET].values.astype(np.float64), np.nan)

    for model_name, pred_log1p in preds.items():
        pred_raw = np.maximum(np.expm1(pred_log1p.astype(np.float64)), 0.0)
        m = evaluate(true_raw[outage_obs], pred_raw[outage_obs])
        seg_outage_rows.append({"model": model_name, "SMAPE": m["SMAPE"], "R2": m["R2"]})

seg_outage = pd.DataFrame(seg_outage_rows).groupby("model").median().loc[BASELINE_NAMES]

print("\nSegment SMAPE comparison (median across entities):")
compare = pd.DataFrame({
    "Weekday"  : seg_weekday["SMAPE"],
    "Weekend"  : seg_weekend["SMAPE"],
    "Outage"   : seg_outage["SMAPE"],
    "Full Test": df_results.groupby("model")["test_SMAPE"].median().loc[BASELINE_NAMES],
}).round(2)
print(compare.to_string())
```

## 7. Run on Institution Subnets (Secondary Track)

```python
panel_sub = pd.read_parquet(
    PREPROCESSED_ROOT / "panel_institution_subnets.parquet",
    columns=["id_institution_subnet", "id_time",
             TARGET, f"{TARGET}_log1p", f"{TARGET}_scaled",
             "observed_mask", "split", "is_outage"]
)
panel_sub = panel_sub.sort_values(["id_institution_subnet", "id_time"]).reset_index(drop=True)

with open(PREPROCESSED_ROOT / "scalers" / "scalers_institution_subnets.pkl", "rb") as f:
    scalers_sub = pickle.load(f)

print(f"Subnet panel: {panel_sub.shape}")
print(f"  Entities: {panel_sub.id_institution_subnet.nunique()}")
```

```python
sub_results = []
for eid, grp in tqdm(panel_sub.groupby("id_institution_subnet"), desc="Subnets"):
    grp    = grp.set_index("id_time").sort_index()
    preds  = make_predictions_entity(grp[f"{TARGET}_log1p"], grp["observed_mask"])
    scaler = scalers_sub.get(eid)
    if scaler is None:
        continue

    full_obs   = grp["observed_mask"].values
    test_obs   = (grp.index >= VAL_END) & (full_obs == 1)
    outage_idx = (grp.index >= OUTAGE_START_ID) & (grp.index <= OUTAGE_END_ID)
    normal_obs = test_obs & ~outage_idx
    true_raw   = np.where(full_obs == 1, grp[TARGET].values.astype(np.float64), np.nan)

    for model_name, pred_log1p in preds.items():
        pred_raw = np.maximum(np.expm1(pred_log1p.astype(np.float64)), 0.0)
        m_test   = evaluate(true_raw[test_obs],   pred_raw[test_obs])
        m_normal = evaluate(true_raw[normal_obs], pred_raw[normal_obs])
        sub_results.append({
            "id_institution_subnet": eid,
            "model"        : model_name,
            "test_SMAPE"   : m_test["SMAPE"],
            "test_R2"      : m_test["R2"],
            "test_RMSE"    : m_test["RMSE"],
            "normal_SMAPE" : m_normal["SMAPE"],
            "normal_R2"    : m_normal["R2"],
        })

df_sub = pd.DataFrame(sub_results)

print("\nSubnet results — median across entities (outage excluded):")
sub_agg = df_sub.groupby("model")[["normal_SMAPE", "normal_R2"]].median().loc[BASELINE_NAMES]
print(sub_agg.round(4).to_string())
```

### Đọc từ preprocessed-statistics-report.md

| Insight | Implication cho baseline |
|---|---|
| Weekday p50=143M vs Weekend p50=32M (**4.4× drop**) | SN-168 quan trọng — bắt được weekly pattern; Mean và MA-24 fail trên weekend |

| Institution 31: max_gap=2h, std cao | Baseline R² thấp hơn → entity khó dự báo |

### Kết quả kỳ vọng
- **SN-168** là baseline tốt nhất (bắt daily+weekly seasonality)
- **MA-168** thường tương đương hoặc kém hơn SN-168 một chút (smoothed over)
- **Mean** → R² ≈ 0 by definition (predict constant = mean)
- **Outage** → tất cả baselines đều tệ hơn vì traffic drop bất ngờ không thể dự báo từ pattern lịch sử
